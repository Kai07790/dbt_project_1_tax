```python
import pandas as pd
```

```python
#원본데이터 인코딩 형식 확인용
from charset_normalizer import from_path

files = [
    "../raw_data/national_tax_raw.csv",
    "../raw_data/local_tax_metro_raw.csv",
    "../raw_data/local_tax_provincial_raw.csv",
    "../raw_data/population_raw.csv",
]

for file in files:
    result = from_path(file).best()

    print(
        file,
        "->",
        result.encoding if result else "unknown"
    )
```

```python
# 1. 국세 데이터 처리
df_nat = pd.read_csv('../raw_data/national_tax_raw.csv', encoding='utf-8', header=[0, 1, 2, 3])

df_nat_melted = df_nat.melt(
    id_vars=df_nat.columns[:6].tolist(), 
    var_name=['연도', '지방청', '지역', '금액 단위'], 
    value_name='세수액'
)

df_nat_melted.columns = [
    '세목별1',
    '세목별2',
    '세목별3',
    '세목별4',
    '세목별5',
    '세목별6',
    '연도',
    '지방청',
    '지역',
    '금액_단위',
    '세수액'
]

df_nat_melted.to_csv('../processed_data/national_tax_processed.csv', index=False, encoding='utf-8')

print(df_nat_melted.columns)
print(df_nat_melted.head())
```

```python
# 2. 지방세 데이터 처리 (특별시, 광역시)
df_met = pd.read_csv('../raw_data/local_tax_metro_raw.csv', encoding='utf-8', header=[0, 1, 2])

df_met_melted = df_met.melt(
    id_vars=df_met.columns[:1].tolist(), 
    var_name=['연도', '분류', '세목'], 
    value_name='세수액'
)

df_met_melted.columns = [
    '자치단체별(1)',
    '연도',
    '분류',
    '세목',
    '세수액'
]

df_met_melted.to_csv('../processed_data/local_tax_metro_processed.csv', index=False, encoding='utf-8')

print(df_met_melted.columns)
print(df_met_melted.head())
```
